```python
# Binomial Distribution
from scipy.stats import binom

# Parameters
n = 10  # Number of trials
p = 0.5 # Probability of success (e.g., a fair coin flip)

# Calculate the probability of exactly 5 successes (heads)

k = 5
prob_5_heads = binom.pmf(k, n, p) #pmf= probability mass (density) function
print(f"Probability of exactly {k} heads: {prob_5_heads:.4f}")

# Calculate the probability of 4 or fewer successes (heads); cumulative binomial distribution function
prob_le_4_heads = binom.cdf(4, n, p)
print(f"Probability of 4 or fewer heads: {prob_le_4_heads:.4f}")

# Calculate the mean and variance given the properties of the binomial distribution formula
mean = n * p
variance = n * p * (1 - p)
print(f"Mean: {mean}, Variance: {variance}")

# Calculate the mean and variance using built-in scipy functions
mean, variance = binom.stats(n, p, moments='mv')
print(f"Mean via moments: {mean}, Variance via moments: {variance}")
```

```python
# Poisson Distribution

# Poisson distribution example from class
# On average, there are 2 alligators per km^2 in the Everglades. What is the probability that given a km^2 contains 4 alligators?
lmbda = 2  # Average rate (alligators per km^2); in python, lambda is a function, so we use lmbda or alternatives instead
k = 4      # Number of alligators we want the probability for

# Poisson formula: (lambda^k * e^-lambda) / k!
probability = (math.pow(lmbda, k) * math.exp(-lmbda)) / math.factorial(k)

print(f"The probability of finding {k} alligators is: {probability:.4f}")
```

```python
from scipy.stats import poisson

# Define the mean number of events (λ)
mu_value = 3 #remember that the mean value of this distribution is E(X) = mu = lambda

# Probability Mass/Denisty Function (PMF)
# Calculate the probability of exactly 4 events occurring (P(X = 4))
probability_of_4 = poisson.pmf(k=4, mu=mu_value)
print(f"Probability of exactly 4 events: {probability_of_4}")

# Cumulative Distribution Function (CDF)
# Calculate the probability of 2 or fewer events occurring (P(X <= 2))
probability_le_2 = poisson.cdf(k=2, mu=mu_value)
print(f"Probability of 2 or fewer events: {probability_le_2}")

# Calculate the probability of more than 4 events occurring (P(X > 4))
# This is 1 - P(X <= 4)
probability_gt_4 = 1 - poisson.cdf(k=4, mu=mu_value)
print(f"Probability of more than 4 events: {probability_gt_4}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# use our example from xearlier
n = 10  # Number of trials
p = 0.5 # Probability of success (e.g., a fair coin flip)
k = np.arange(0, n + 1) # list of possible outcomes

# Compute binomial distribution
pmf = binom.pmf(k, n, p)

# Plot Histogram
plt.figure()
plt.bar(k, pmf)
plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
```

```python
import scipy.stats as stats

# Given values
n = 20
xbar = 9.5
s2 = 5.10
s = np.sqrt(s2)
df = n - 1
alpha = 0.05

# Mean Confidence Intervals
t_crit = stats.t.ppf(1 - alpha/2, df)
SE = s / np.sqrt(n)
mean_lower = xbar - t_crit * SE
mean_upper = xbar + t_crit * SE

# Variance Confidence Intervals
chi2_lower = stats.chi2.ppf(alpha/2, df)
chi2_upper = stats.chi2.ppf(1 - alpha/2, df)

var_lower = (df * s2) / chi2_upper
var_upper = (df * s2) / chi2_lower

# Standard Deviation Confidence Intervals
std_lower = np.sqrt(var_lower)
std_upper = np.sqrt(var_upper)

print("95% CI for mean:", (mean_lower, mean_upper))
print("95% CI for variance:", (var_lower, var_upper))
print("95% CI for std dev:", (std_lower, std_upper))
```
